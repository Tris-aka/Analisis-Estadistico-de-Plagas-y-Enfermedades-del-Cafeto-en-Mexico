```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("Plagas_limpio.csv", encoding="utf-8-sig")

plagas = [
    "Mancha Hierro",
    "Ojo de Gallo",
    "Phoma",
    "Minador",
    "Nematodo Lesionador",
]
```

```python
# Aporte (%) de cada plaga al Factor de Daño por Estado
aporte_pct = (
    df.groupby("Estado")[plagas]
    .sum()
    .div(df.groupby("Estado")["Factor Danio"].sum(), axis=0)
    * 100
)
print("\nAporte (%) por Estado:")
print(aporte_pct.round(2))
```

```python
# Aporte (%) por Estado
plt.figure(figsize=(8, 4.5))
aporte_pct.plot(
    kind="barh", stacked=True, ax=plt.gca(), colormap="viridis", width=0.6
)
plt.title("Aporte (%) de cada plaga al Factor de Daño por Estado")
plt.xlabel("Porcentaje de aporte (%)")
plt.xlim(0, 100)
plt.legend(title="Plaga", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
for estado in ["Chiapas", "Puebla", "Veracruz"]:
  print(f" {estado}")
  print(df[df["Estado"] == estado]["Factor Danio"].describe().round(4))
  print()
```

```python
# Boxplot de Factor de Daño por Estado
plt.figure(figsize=(7, 4.5))
df.boxplot(column="Factor Danio", by="Estado", ax=plt.gca())
plt.suptitle("")
plt.title("Boxplot de Factor de Daño por Estado")
plt.xlabel("Estado")
plt.ylabel("Factor de Daño")
plt.tight_layout()
plt.show()
```
